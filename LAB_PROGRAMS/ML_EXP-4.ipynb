```python
import numpy as np

# XOR training dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

Y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Initialize weights and biases
np.random.seed(42)

W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))

W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Derivative of sigmoid
def sigmoid_derivative(x):
    return x * (1 - x)

# Train using backpropagation
learning_rate = 0.5
epochs = 10000

for epoch in range(epochs):

    # Forward propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    output = sigmoid(final_input)

    # Calculate error
    error = Y - output

    # Backpropagation
    d_output = error * sigmoid_derivative(output)
    error_hidden = d_output.dot(W2.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_output)

    # Update weights and biases
    W2 += learning_rate * hidden_output.T.dot(d_output)
    b2 += learning_rate * np.sum(d_output, axis=0, keepdims=True)

    W1 += learning_rate * X.T.dot(d_hidden)
    b1 += learning_rate * np.sum(d_hidden, axis=0, keepdims=True)

    if (epoch + 1) % 2000 == 0:
        print("Epoch:", epoch + 1,
              "Loss:", round(np.mean(error ** 2), 6))

# Test the trained neural network
print("\nTesting the Neural Network:")
print("Input\tPredicted Output")

for x in X:
    hidden = sigmoid(np.dot(x.reshape(1, -1), W1) + b1)
    result = sigmoid(np.dot(hidden, W2) + b2)

    predicted = 1 if result[0, 0] >= 0.5 else 0

    print(x, "\t", predicted)

# Test a new sample
sample = np.array([[0, 1]])

hidden = sigmoid(np.dot(sample, W1) + b1)
result = sigmoid(np.dot(hidden, W2) + b2)

print("\nNew Input:", sample[0])
print("Predicted Probability:", round(result[0, 0], 4))
print("Predicted Class:", int(result[0, 0] >= 0.5))
```
