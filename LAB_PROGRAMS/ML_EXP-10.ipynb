```python
import numpy as np
from sklearn.mixture import GaussianMixture

# Sample dataset
X = np.array([
    [1.0], [1.5], [2.0], [2.5], [3.0],
    [8.0], [8.5], [9.0], [9.5], [10.0]
])

print("Training Dataset:")
print(X.ravel())

# Create the Gaussian Mixture Model
gmm = GaussianMixture(
    n_components=2,
    random_state=42,
    n_init=5
)

# Apply Expectation-Maximization
gmm.fit(X)

# Predict cluster assignments
labels = gmm.predict(X)

print("\nCluster Assignments:")
for value, label in zip(X.ravel(), labels):
    print(value, "-> Cluster", label)

# Display learned parameters
print("\nCluster Means:")
print(gmm.means_.ravel().round(2))

print("\nCluster Weights:")
print(gmm.weights_.round(2))

print("\nNumber of EM Iterations:", gmm.n_iter_)

# Predict cluster for a new sample
sample = np.array([[2.2]])
prediction = gmm.predict(sample)

print("\nNew Sample:", sample[0][0])
print("Predicted Cluster:", prediction[0])

# Display cluster probabilities
probabilities = gmm.predict_proba(sample)
print("Cluster Probabilities:", probabilities.round(3))
```
