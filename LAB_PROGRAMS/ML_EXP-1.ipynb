```python
import pandas as pd

# =========================================================
# DATASET: Classic "Enjoy Sport" Training Examples
# Columns: Sky, AirTemp, Humidity, Wind, Water, Forecast, EnjoySport
# =========================================================
dataset = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same',   'Yes'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Warm', 'Same',   'Yes'],
    ['Rainy', 'Cold', 'High',   'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Cool', 'Change', 'Yes'],
]

columns = ['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast', 'EnjoySport']


# =========================================================
# FIND-S ALGORITHM
# =========================================================
def find_s(training_data):
    """
    Implements the FIND-S algorithm to find the most specific hypothesis
    consistent with all positive training examples.
    """
    # Step 1: Initialize hypothesis with the FIRST positive example
    hypothesis = None
    for example in training_data:
        if example[-1] == 'Yes':
            hypothesis = list(example[:-1])
            break

    if hypothesis is None:
        return "No positive examples found in the dataset"

    print(f"\nStep 1: Initialize hypothesis with first positive example:")
    print(f"   h0 = {hypothesis}\n")

    print("Step 2: Compare h with each training example:\n")
    # Step 2: Iterate through all examples
    for i, example in enumerate(training_data, start=1):
        attributes = example[:-1]
        target = example[-1]

        if target == 'Yes':
            # Compare and generalize
            changed = False
            for j in range(len(hypothesis)):
                if hypothesis[j] != attributes[j]:
                    hypothesis[j] = '?'
                    changed = True

            if changed:
                print(f"   Example {i}: {example}")
                print(f"   -> Positive, MISMATCH -> generalize")
                print(f"   h{i} = {hypothesis}\n")
            else:
                print(f"   Example {i}: {example}")
                print(f"   -> Positive, matches hypothesis (no change)")
                print(f"   h{i} = {hypothesis}\n")
        else:
            print(f"   Example {i}: {example}")
            print(f"   -> Negative example, IGNORED by FIND-S\n")

    return hypothesis


# =========================================================
# MAIN EXECUTION
# =========================================================
if __name__ == "__main__":
    print("=" * 75)
    print("           FIND-S ALGORITHM - CONCEPT LEARNING DEMO")
    print("=" * 75)

    # Display dataset
    print("\n>>> TRAINING DATASET (Enjoy Sport):\n")
    df = pd.DataFrame(dataset, columns=columns)
    df.index = [f"Ex{i}" for i in range(1, len(dataset) + 1)]
    print(df.to_string())
    print(f"\nTotal examples: {len(dataset)}")
    print(f"Positive examples: {sum(1 for e in dataset if e[-1] == 'Yes')}")
    print(f"Negative examples: {sum(1 for e in dataset if e[-1] == 'No')}")

    # Run FIND-S
    print("\n" + "=" * 75)
    print("           FIND-S ALGORITHM EXECUTION")
    print("=" * 75)
    final_hypothesis = find_s(dataset)

    # Display result
    print("=" * 75)
    print("           FINAL HYPOTHESIS (Most Specific)")
    print("=" * 75)
    print(f"\n   h = {final_hypothesis}\n")

    print("Interpretation: 'Enjoy Sport = Yes' when:\n")
    for col, val in zip(columns[:-1], final_hypothesis):
        meaning = "any value is acceptable" if val == '?' else f"must be '{val}'"
        print(f"   - {col:<10} = {val:<8} ({meaning})")

    print("\n" + "=" * 75)
    print("  '?' means the value doesn't matter (generalized attribute)")
    print("=" * 75)
```
