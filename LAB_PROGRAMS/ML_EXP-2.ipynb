```python
import pandas as pd

# Create the training dataset
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

# Save data to CSV and read it
pd.DataFrame(data).to_csv('enjoysport.csv', index=False)
df = pd.read_csv('enjoysport.csv')

print("Training Data:")
print(df.to_string(index=False))

X = df.iloc[:, :-1].values
Y = df.iloc[:, -1].values

n = len(X[0])
S = ['0'] * n
G = [['?'] * n]

for i in range(len(X)):
    if Y[i] == 'Yes':
        for j in range(n):
            if S[j] == '0':
                S[j] = X[i][j]
            elif S[j] != X[i][j]:
                S[j] = '?'

        G = [g for g in G if all(
            g[j] == '?' or g[j] == X[i][j]
            for j in range(n)
        )]

    else:
        new_G = []
        for g in G:
            if all(g[j] == '?' or g[j] == X[i][j]
                   for j in range(n)):
                for j in range(n):
                    if S[j] != '?' and S[j] != '0' and S[j] != X[i][j]:
                        h = g.copy()
                        h[j] = S[j]
                        new_G.append(h)
            else:
                new_G.append(g)
        G = new_G

    print("\nTraining Example:", i + 1)
    print("Specific Boundary S:", S)
    print("General Boundary G:", G)

print("\nFinal Specific Boundary:", S)
print("Final General Boundary:")
for g in G:
    print(g)

print("\nCandidate-Elimination completed successfully.")
```
